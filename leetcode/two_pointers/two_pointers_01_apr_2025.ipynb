```python
# 881. Boats to Save People
# Medium
# Topics
# Companies
# You are given an array people where people[i] is the weight of the ith person, and an infinite number of boats where each boat can carry a maximum weight of limit. Each boat carries at most two people at the same time, provided the sum of the weight of those people is at most limit.

# Return the minimum number of boats to carry every given person.

 

# Example 1:

# Input: people = [1,2], limit = 3
# Output: 1
# Explanation: 1 boat (1, 2)
# Example 2:

# Input: people = [3,2,2,1], limit = 3
# Output: 3
# Explanation: 3 boats (1, 2), (2) and (3)
# Example 3:

# Input: people = [3,5,3,4], limit = 5
# Output: 4
# Explanation: 4 boats (3), (3), (4), (5)
```

```python
from typing import List, Optional, Tuple

def numRescueBoats(people: List[int], limit: int) -> int:
    people.sort()
    n = len(people)
    l = 0
    r = n -1 
    boat_count = 0
    while l <= r:
        boat_count += 1
        if l == r:
            print(f"Last boat will have 1 person")
            l = l+ 1
            continue
        
        max_weight = people[r]
        r = r -1
        rem_weight = limit - max_weight
        if people[l] <= rem_weight:
            print(f"Boat {boat_count} will have {people[l]} and {max_weight}")
            l += 1
        else:
            print(f"Boat {boat_count} will have {max_weight}")
    return boat_count


```

```python
numRescueBoats(people = [2], limit = 4)
```

```python
numRescueBoats(people = [3,2,2,1], limit = 3)
```

```python
# 189. Rotate Array
# Medium
# Topics
# Companies
# Hint
# Given an integer array nums, rotate the array to the right by k steps, where k is non-negative.

 

# Example 1:

# Input: nums = [1,2,3,4,5,6,7], k = 3
# Output: [5,6,7,1,2,3,4]
# Explanation:
# rotate 1 steps to the right: [7,1,2,3,4,5,6]
# rotate 2 steps to the right: [6,7,1,2,3,4,5]
# rotate 3 steps to the right: [5,6,7,1,2,3,4]
# Example 2:

# Input: nums = [-1,-100,3,99], k = 2
# Output: [3,99,-1,-100]
# Explanation: 
# rotate 1 steps to the right: [99,-1,-100,3]
# rotate 2 steps to the right: [3,99,-1,-100]
```

```python
def rotate(nums:List[int], k:int) -> List[int]:
    n = len(nums)
    rotated = [None] * n
    k_div, k_rem = divmod(k,n)
    if k_rem == 0:
        return nums
    rotated[0:k_rem] = nums[-k_rem:]
    rotated[k_rem:] = nums[:(n-k_rem)]
    return rotated
    
```

```python
rotate(nums = [1,2,3,4,5,6,7], k = 3)
```

```python
rotate([-1,-100,3,99], k = 2)
```

```python
# 88. Merge Sorted Array
# Easy
# Topics
# Companies
# Hint
# You are given two integer arrays nums1 and nums2, sorted in non-decreasing order, and two integers m and n, representing the number of elements in nums1 and nums2 respectively.

# Merge nums1 and nums2 into a single array sorted in non-decreasing order.

# The final sorted array should not be returned by the function, but instead be stored inside the array nums1. To accommodate this, nums1 has a length of m + n, where the first m elements denote the elements that should be merged, and the last n elements are set to 0 and should be ignored. nums2 has a length of n.

```

```python
def merge(nums1: List[int], nums2: List[int]) -> List[int]:
    n_nums2 = len(nums2)
    n_nums1 = len(nums1) - n_nums2
    #print(n_nums1, n_nums2)
    nums1_ind = n_nums1 -1
    nums2_ind = n_nums2 - 1
    while nums1_ind >= 0 and nums2_ind >= 0:
        print(nums1)
        print(nums1_ind,nums2_ind)
        print(nums1_ind + nums2_ind -1)
        print(f"Comparing {nums1[nums1_ind]} with {nums2[nums2_ind]}")
        if nums1[nums1_ind] >= nums2[nums2_ind]:
            nums1[nums1_ind + nums2_ind + 1] = nums1[nums1_ind]
            nums1_ind = nums1_ind -1 
        else:
            nums1[nums1_ind + nums2_ind + 1] = nums2[nums2_ind]
            nums2_ind = nums2_ind -1
    return nums1
```
