```python
# Trapping Rain Water
# You are given an array non-negative integers height which represent an elevation map. Each value height[i] represents the height of a bar, which has a width of 1.

# Return the maximum area of water that can be trapped between the bars.

# Example 1:



# Input: height = [0,2,0,3,1,0,1,3,2,1]

# Output: 9
```

```python
def trap_water(height:List[int]) -> int:
    n = len(height)
    leftMaxStore = [-1] * n
    rightMaxStore = [-1] * n
    lm, rm = -1, -1
    for i in range(n):
        ## FIll left max and update lm
        leftMaxStore[i] = lm
        lm = max(lm, height[i])
        
        ##  fill right max and update rm
        rightMaxStore[(n-1) - i] = rm
        rm = max(rm, height[(n-1) - i])
    
    water_stored = [min(i,j) - h for i, j, h in zip(leftMaxStore,rightMaxStore, height)]
    
    return sum([w for w in water_stored if w>= 0])
    
        
```

```python
trap_water(height = [0,2,0,3,1,0,1,3,2,1])
```

```python
trap_water([0,1,0,2,1,0,1,3,2,1,2,1])
```
